```python
import torch
import torch.nn as nn

class TokenEmbeddings(nn.Module):
    
    def __init__(self,vocab_size,num_dims):
        super().__init__()
        self.embeddings = nn.Parameter(torch.randn(vocab_size,num_dims))

    def forward(self,indices):
        return self.embeddings(indices)

class PositionalEmbeddings(nn.Module):

    def __init__(self,seq_len,num_dims):
        super().__init__()
        self.weights = nn.Parameter(torch.randn(seq_len, num_dims))

    def forward(self,positions):
        return self.weights[positions]

```
